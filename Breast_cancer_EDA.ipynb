```python
# Feature Engineering:
bins_age = [0, 40, 60, df['Age'].max() + 1]
labels_age = ['Young (<40)', 'Middle-Aged (40-60)', 'Elderly (>60)']
df['Age_Group'] = pd.cut(df['Age'], bins=bins_age, labels=labels_age)
```

```python

bins_tumor = [0, 20, 50, df['Tumor Size'].max() + 1]
labels_tumor = ['Small (<=20mm)', 'Medium (20-50mm)', 'Large (>50mm)']
df['Tumor_Size_Category'] = pd.cut(df['Tumor Size'], bins=bins_tumor, labels=labels_tumor)
```

```python
#univariate analysis

sns.set_theme(style="whitegrid")
plt.figure(figsize=(15, 12))


plt.subplot(2, 2, 1)
sns.histplot(df['Age'], kde=True, color='teal', bins=30)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
```

```python
#bivariate analysis

sns.set_theme(style="whitegrid", font_scale=1.1)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Age vs Tumor Size
sns.scatterplot(
    data=df,
    x='Age',
    y='Tumor Size',
    hue='Status',
    alpha=0.6,
    palette='Set1',
    ax=axes[0, 0]
)
axes[0, 0].set_title('Age vs Tumor Size (by Status)', fontweight='bold')

# 2. Age Distribution by Status
sns.boxplot(
    data=df,
    x='Status',
    y='Age',
    palette='crest',
    ax=axes[0, 1]
)
axes[0, 1].set_title('Age Distribution by Status', fontweight='bold')

# 3. Tumor Size by Status
sns.violinplot(
    data=df,
    x='Status',
    y='Tumor Size',
    palette='mako',
    inner='quartile',
    ax=axes[1, 0]
)
axes[1, 0].set_title('Tumor Size Distribution by Status', fontweight='bold')

# 4. Survival Months by Status
sns.kdeplot(
    data=df,
    x='Survival Months',
    hue='Status',
    fill=True,
    common_norm=False,
    palette='Set1',
    ax=axes[1, 1]
)
axes[1, 1].set_title('Survival Months Density by Status', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[25, 45, 60, 75],
    labels=['Young (<45)', 'Middle (45-60)', 'Senior (60+)']
)

stage_order = ['IIA', 'IIB', 'IIIA', 'IIIB', 'IIIC']

plt.figure(figsize=(11, 6))

sns.boxplot(
    data=df,
    x='6th Stage',
    y='Survival Months',
    hue='Age Group',
    order=stage_order,
    palette='Blues'
)

plt.title('Survival Months by 6th Stage across Age Groups', fontweight='bold', fontsize=13)
plt.xlabel('6th Stage', fontsize=11)
plt.ylabel('Survival Months', fontsize=11)
plt.legend(title='Age Group', bbox_to_anchor=(1.02, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
#time series analysis

# Age Groups
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[29, 39, 49, 59, 69],
    labels=['30-39', '40-49', '50-59', '60-69']
)

# Single plot canvas
plt.figure(figsize=(8, 5))

age_counts = df['Age Group'].value_counts().sort_index()
sns.barplot(x=age_counts.index, y=age_counts.values, palette='Blues_r')

plt.title('Age group by chances of cancer', fontsize=13, fontweight='bold')
plt.xlabel('Age Group (Years)', fontsize=11)
plt.ylabel('Total Number of Patients', fontsize=11)

for i, v in enumerate(age_counts.values):
    plt.text(i, v + 25, f"{v} cases\n({v/len(df)*100:.1f}%)", ha='center', fontweight='bold', fontsize=10)

plt.ylim(0, 1600)
plt.tight_layout()
plt.show()
```

***Key Insights***


Based on the exploratory data analysis and visualizations, here are the key insights from the dataset:

**Age Distribution:** The majority of patients are concentrated in the middle-aged and older cohorts, specifically between 40 and 69 years of age. The 50-59 age group has the highest occurrence of cases (34.6%), closely followed by the 60-69 age group (31.8%).

**Survival and Cancer Grade:** There is a clear relationship between the tumor grade and survival months. Patients with lower grades (Grade 1) show higher survival times compared to those diagnosed with higher grades (Grade 3 or Grade 4/anaplastic), where survival months drop significantly, particularly for deceased patients.

**Tumor Size and Survival:** Larger tumor sizes generally correlate with lower survival times and a higher mortality rate. Most surviving patients (labeled Alive) are clustered at longer survival timelines with smaller tumor sizes, whereas those labeled Dead show higher density in shorter survival periods.

**Node Positivity and Stages:** As the cancer stage advances from IIA to IIIC, the number of regional positive lymph nodes increases dramatically. Stage IIIC patients exhibit the highest average number of positive nodes, which correlates with more aggressive disease progression.

**Correlations:** There is a moderate positive correlation (0.41) between the number of regional nodes examined and those found positive. However, other numeric factors like age show very low direct linear correlation with tumor size or survival months, suggesting survival outcomes are determined by complex multi-variable interactions rather than age alone.
